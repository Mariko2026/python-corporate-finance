```python
#Affichage des indicateurs
chiffre_affaires = 500000
cout_reel = 380000
budget_previsionnel = 350000

#Calcul de la marge
marge = chiffre_affaires - cout_reel

#Calcul du taux de marge sur le chiffre d'affaires
taux_de_marge = (marge / chiffre_affaires) *100

#Ecart budgétaire
ecart = cout_reel - budget_previsionnel

#Ecart en pourcentage
ecart_en_pourcentage = (ecart / budget_previsionnel) * 100


#Diagnostic
if cout_reel > budget_previsionnel:
  print("Alerte, budget dépassé !")
elif cout_reel == budget_previsionnel:
  print("Budget exactement respecté")
else:
  print("Economie réalisée")

#Affichage des résultats
print("Chiffre d'affaires :", chiffre_affaires, "€")
print("Coût réel :", cout_reel, "€")
print("Budget prévisionnel :", budget_previsionnel, "€")
print("Marge :", marge, "€")
print("Taux de marge :", round(taux_de_marge,2), "%")
print("Ecart :", ecart, "€")
print("Ecart en pourcentage :", round(ecart_en_pourcentage,2), "%")
```
